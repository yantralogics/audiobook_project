# Audiobook TTS Workflow Demo

This notebook demonstrates the intended workflow: load the audiobook PRD, build prompts for a 30-second synopsis and a full story, generate audio from text-to-speech, and play the results inline.


```python
from pathlib import Path
from dotenv import load_dotenv
from IPython.display import Audio, display
import importlib
import whisper_client
importlib.reload(whisper_client)
from whisper_client import generate_audiobook_audio

load_dotenv()

project_root = Path('.')
prd_file = project_root / 'audiobook_prd.md'
sample_audio_file = project_root / 'sample_synopsis_audio.mp3'
full_audio_file = project_root / 'full_story_audio.mp3'

print(f'Using PRD file: {prd_file}')
```

## Prompt generation helpers

Define the text prompts for the sample synopsis and the full audiobook story. The PRD is automatically included when generating audio.


```python
def make_sample_prompt() -> str:
    return (
        'Create a 30-second synopsis for a warm, whimsical audiobook for Meera. '
        'Describe the story arc, narration style, and gentle mood. '
        'Mention Meera as a bright-eyed, happy, goofy girl who loves coloring, playing doctor with stuffies, cats, and unicorns. '
        'Include natural elements, discovery, curiosity, bravery, and problem solving. '
        'Keep the language simple and nurturing.'
    )

def make_full_story_prompt() -> str:
    return (
        'Create a full audiobook story for Meera with a clear beginning, middle, and end. '
        'Use warm, calm, expressive narration and gentle pacing. '
        'Include her friends Anna, Callie, Sam, Gannon, Luke, and teachers Kim and Sitwa. '
        'Build the story around natural discovery, curiosity, bravery, and problem solving. '
        'Keep vocabulary appropriate for a 4-year-old and aim for a gentle listening experience.'
    )
```

## Generate sample synopsis audio

Create a short preview audio file using the 30-second synopsis prompt plus the PRD context.


```python
sample_prompt = make_sample_prompt()
print('Sample prompt:')
print(sample_prompt)

sample_audio_path = generate_audiobook_audio(
    prd_file_path=str(prd_file),
    prompt=sample_prompt,
    output_file_path=str(sample_audio_file),
    model='gpt-4o-mini-tts',
    voice='marin',
    response_format='mp3',
    instructions='Narrate in a warm, gentle storyteller voice.',
)
print(f'Generated sample audio: {sample_audio_path}')
```

## Generate full story audio

Create the full audiobook audio file from the full story prompt plus PRD context.


```python
full_story_prompt = make_full_story_prompt()
print('Full story prompt:')
print(full_story_prompt)

full_audio_path = generate_audiobook_audio(
    prd_file_path=str(prd_file),
    prompt=full_story_prompt,
    output_file_path=str(full_audio_file),
    model='gpt-4o-mini-tts',
    voice='marin',
    response_format='mp3',
    instructions='Narrate in a warm, calm, expressive voice with gentle pacing.',
)
print(f'Generated full story audio: {full_audio_path}')
```

## Playback generated audio

If generation completed successfully, play the created MP3 files directly in the notebook.


```python
for audio_path in [sample_audio_file, full_audio_file]:
    if audio_path.exists():
        print(f'Playing {audio_path.name}')
        display(Audio(filename=str(audio_path)))
    else:
        print(f'Missing audio file: {audio_path}')
```
